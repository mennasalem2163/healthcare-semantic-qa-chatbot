```python
import pandas as pd

def generate_ngrams(text, n):
    # Check if the text is a string
    if isinstance(text, str):
        # Split text into words/numbers (tokens)
        tokens = text.split()

        # Generate n-grams
        ngrams = zip(*[tokens[i:] for i in range(n)])

        # Convert tuples to 'n' tokens
        return [" ".join(ngram) for ngram in ngrams]
    else:
        # Handle non-string text (e.g., NaN or numbers)
        return []

# Process each text entry to generate n-grams
for text in df['answer']:
    ngrams = generate_ngrams(text, n)
    print(f"Text: {text}")
    print(f"Generated {n}-grams: {ngrams}\n")
```

```python
import nltk
nltk.download('wordnet')
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

# Define the function for cleaning text
def clean_text(text):
    if not isinstance(text, str):
        return ""
    text = re.sub(r'[^a-zA-Z\s]', '', text.lower())
    return text

# Initialize the lemmatizer
lemmatizer = WordNetLemmatizer()

# Define the function for preprocessing text
def preprocess_text(text, verbose=False):
    if verbose: print(f"Original text: {text}")
    cleaned = clean_text(text)
    if verbose: print(f"Cleaned text: {cleaned}")

    tokenized = word_tokenize(cleaned)
    if verbose: print(f"Tokenized text: {tokenized}")

    # Applying Lemmatization
    lemmatized = [lemmatizer.lemmatize(word) for word in tokenized]
    if verbose: print(f"Lemmatized text: {lemmatized}")

    # Removing stopwords
    stop_words = set(stopwords.words('english'))
    no_stopwords = [word for word in lemmatized if word not in stop_words]
    if verbose: print(f"Text without stopwords: {no_stopwords}\n")

    return no_stopwords

# Apply preprocessing to the 'question' and 'answer' columns
df['processed_question'] = df['question'].apply(preprocess_text)
df['processed_answer'] = df['answer'].apply(preprocess_text)

# Test with verbose output
for i, row in df.head(5).iterrows():
    print(f"Row {i}:")
    _ = preprocess_text(row['question'], verbose=True)
    _ = preprocess_text(row['answer'], verbose=True)
```

```python
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
from transformers import pipeline
from sentence_transformers import SentenceTransformer


# Initialize the sentence transformer model for semantic matching
model = SentenceTransformer('all-MiniLM-L6-v2')

# Initialize the summarization pipeline
summarizer = pipeline("summarization")

def find_similar_question(user_question, questions):
    question_embeddings = model.encode(questions)
    user_question_embedding = model.encode([user_question])
    similarities = cosine_similarity(user_question_embedding, question_embeddings)
    most_similar_idx = similarities[0].argmax()
    return questions[most_similar_idx]

def summarize_answer(answer):
    summary = summarizer(answer, max_length=50, min_length=25, do_sample=False)
    return summary[0]['summary_text']

def find_answer(question, df):
    questions = df['question'].tolist()
    matched_question = find_similar_question(question, questions)
    matches = df[df['question'] == matched_question]

    if not matches.empty:
        return summarize_answer(matches.iloc[0]['answer'])
    else:
        return "Sorry, I don't have an answer for that."

def find_disease(answer, df):
    matches = df[df['answer'].str.contains(answer, case=False, na=False)]
    if not matches.empty:
        return matches.iloc[0]['disease']
    else:
        return "Sorry, I can't find a disease related to that answer."

def chatbot(df):
    print("Hello! How can I help you? Type 'bye', 'goodbye', 'quit', or 'exit' to end the chat.")

    while True:
        user_input = input("Enter your question or answer: ")

        if user_input.lower() in ['bye', 'goodbye', 'quit', 'exit']:
            print("Goodbye! Have a great day!")
            break

        if "?" in user_input:
            print("Answer:", find_answer(user_input, df))
        else:
            print("Disease:", find_disease(user_input, df))

if __name__ == "__main__":

    chatbot(df)
```
